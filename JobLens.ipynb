filling the numerical columns

```python
df['salary_min_lpa']=df['salary_min_lpa'].astype("object")
```

```python
df['salary_max_lpa']=df['salary_max_lpa'].astype("object")
```

```python
df.dtypes
```

```python
num_colls = df.select_dtypes(include = "number").columns
num_colls
```

```python
num_colls
```

filling numerical columns with meadian

```python
df[num_colls] = df[num_colls].fillna(df[num_colls].median())
```

```python
df.isna().sum()
```

```python
df['salary_max_lpa']=df['salary_max_lpa'].astype("float")
```

```python
df['salary_min_lpa']=df['salary_min_lpa'].astype("float")
```

filling categorical columns with mode

```python
cat_cols = df.select_dtypes(include = "object").columns
```

```python
cat_cols
```

```python
df[cat_cols] = df[cat_cols].fillna(df[cat_cols].mode().iloc[0])
```

```python
df.isna().sum()
```

**outliers**

```python
Q3 = df["salary_min_lpa"].quantile(0.75)
Q1 = df["salary_min_lpa"].quantile(0.25)
IQR = Q3 - Q1
IQR
```

```python
low_limit = Q1 - 1.5*IQR
upp_limit = Q3 + 1.5*IQR
low_limit , upp_limit
```

```python
df = df[(df["salary_min_lpa"] >=low_limit) & (df["salary_min_lpa"]<=upp_limit)]
```

```python
df.shape
```

```python
Q3 = df["salary_max_lpa"].quantile(0.75)
Q1 = df["salary_max_lpa"].quantile(0.25)
IQR = Q3 - Q1
IQR
```

```python
low_limit = Q1 - 1.5*IQR
upp_limit = Q3 + 1.5*IQR
low_limit , upp_limit
```

```python
df = df[(df["salary_max_lpa"] >=low_limit) & (df["salary_max_lpa"]<=upp_limit)]
```
